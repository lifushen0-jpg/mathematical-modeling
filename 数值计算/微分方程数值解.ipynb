## 使用scipy进行微分方程的数值求解

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

# 定义求解的函数
def f(x, y):
    return 2 * y

# 取样点，计算这些点上的 y 的值
x_eval = np.linspace(0, 2, 100)


# 求解
sol = solve_ivp(
    f,                # 所求函数
    t_span=(0, 2),    # x 的范围，是一个闭区间
    y0=[1],           # 初始 x 的值，这里由于 x 的范围为 [0,2]，所以表示 y(0)=1
    t_eval=x_eval     # 取样点
)

plt.plot(sol.t, sol.y[0])
plt.xlabel('x')
plt.ylabel('y')
plt.grid()
plt.show()
```
