## 算法思想  
- 一、隶属函数：将评价指标映射到区间$[0,1]$中，越大越优，其结果称为隶属度
- 二、隶属函数的确定：
1. 模糊统计法(基本不用)
2. 指派法
3. 其他

本算法主观性很强，应用不多  
应该注意的是，本算法的最初目的是topsis发一样，给不同方案进行综合评价，但是也可用于确定不同指标的权重以配合其他算法使用

#### 一、矩形

```python
# 偏小型
def rectangle_small(x, a):
    """偏小型矩阵隶属函数"""
    x = np.asarray(x)
    return np.where(x <= a, 1, 0)
```

```python
# 中间型
def rectangle_middle(x, a, b):
    """
    中间型矩阵隶属函数
    """
    x = np.asarray(x)
    return np.where((x >= a) & (x <= b), 1, 0)
```

```python
# 偏大型
def rectangle_large(x, a):
    """
    偏大型矩阵隶属函数
    """
    x = np.asarray(x)
    return np.where(x >= a, 1, 0)
```

#### 二、梯形

```python
def trapezoid_small(x, a, b):
    """
    偏小梯形隶属函数
    """
    x = np.asarray(x)

    mu = np.zeros_like(x, dtype=float)

    mu[x <= a] = 1

    index = (x > a) & (x <= b)
    mu[index] = (b - x[index]) / (b - a)

    return mu
```

```python
def trapezoid_middle(x, a, b, c, d):
    """
    中间梯形隶属函数
    """

    x = np.asarray(x)

    mu = np.zeros_like(x, dtype=float)

    index = (x >= a) & (x < b)
    mu[index] = (x[index]-a)/(b-a)

    index = (x >= b) & (x <= c)
    mu[index] = 1

    index = (x > c) & (x <= d)
    mu[index] = (d-x[index])/(d-c)

    return mu
```

```python
def trapezoid_large(x, a, b):
    """
    偏大梯形隶属函数
    """ 
    x = np.asarray(x)

    mu = np.zeros_like(x, dtype=float)

    index = (x >= a) & (x < b)
    mu[index] = (x[index]-a)/(b-a)

    mu[x >= b] = 1

    return mu
```

#### 三、k次抛物型

```python
def parabola_small(x, a, b, k):
    """偏小抛物线隶属函数"""
    
    x=np.asarray(x)

    mu=np.zeros_like(x,dtype=float)

    mu[x<=a]=1

    index=(x>a)&(x<=b)

    mu[index]=((b-x[index])/(b-a))**k

    return mu
```

```python
def parabola_middle(x,a,b,c,d,k):
    """中间抛物型隶属函数"""
    x=np.asarray(x)

    mu=np.zeros_like(x,dtype=float)


    index=(x>=a)&(x<b)
    mu[index]=((x[index]-a)/(b-a))**k


    index=(x>=b)&(x<=c)
    mu[index]=1


    index=(x>c)&(x<=d)
    mu[index]=((d-x[index])/(d-c))**k


    return mu
```

```python
def parabola_large(x,a,b,k):
    """
    偏大抛物线隶属函数
    """
    x=np.asarray(x)

    mu=np.zeros_like(x,dtype=float)


    index=(x>=a)&(x<=b)

    mu[index]=((x[index]-a)/(b-a))**k


    mu[x>b]=1


    return mu
```

#### 四、$\Gamma$ 型

```python
def gamma_small(x,a,k):
    """偏小伽马型隶属函数"""
    x=np.asarray(x)

    mu=np.ones_like(x,dtype=float)

    index=x>a

    mu[index]=np.exp(-k*(x[index]-a))

    return mu
```

```python
def gamma_middle(x,a,b,k):
    """中间伽马型隶属函数"""
    x=np.asarray(x)

    mu=np.ones_like(x,dtype=float)


    index=x<a
    mu[index]=np.exp(-k*(a-x[index]))


    index=x>b
    mu[index]=np.exp(-k*(x[index]-b))


    return mu
```

```python
def gamma_large(x,a,k):
    """偏大伽马型隶属函数"""
    x=np.asarray(x)

    mu=np.zeros_like(x,dtype=float)


    index=x>=a

    mu[index]=1-np.exp(-k*(x[index]-a))


    return mu
```

#### 五、正态型

```python
def normal_small(x,a,sigma):
    """偏小正态型隶属函数"""
    x=np.asarray(x)

    mu=np.ones_like(x,dtype=float)

    index=x>a

    mu[index]=np.exp(-((x[index]-a)/sigma)**2)

    return mu
```

```python
def normal_middle(x,a,sigma):
    """中间正态型隶属函数"""
    x=np.asarray(x)

    return np.exp(-((x-a)/sigma)**2)
```

```python
def normal_large(x,a,sigma):
    """偏大正态型隶属函数"""
    x=np.asarray(x)

    mu=np.zeros_like(x,dtype=float)

    index=x>a

    mu[index]=1-np.exp(-((x[index]-a)/sigma)**2)

    return mu
```

#### 六、sigmoid型

```python
def sigmoid_small(x,a,alpha,beta):
    """偏小S型隶属函数"""
    x=np.asarray(x)

    return 1/(1+alpha*np.maximum(x-a,0)**beta)
```

```python
def sigmoid_middle(x,a,alpha,beta):
    """中间S型隶属函数"""
    x=np.asarray(x)

    return 1/(1+alpha*np.abs(x-a)**beta)
```

```python
def sigmoid_large(x,a,alpha,beta):
    """偏大S型隶属函数"""
    x=np.asarray(x)

    return 1-1/(1+alpha*np.maximum(x-a,0)**beta)
```
