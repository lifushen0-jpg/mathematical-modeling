### 我们将使用 scipy 这个库进行非线性规划  
- 代码示例  
问题：  
求  max  4x₁ - 0.3x₁² + 5x₂  
约束为：  
2x₁ + x₂ ≤ 10  
x₁ + 3x₂ ≤ 15  
x₁, x₂ ≥ 0  

```python
import numpy as np
from scipy.optimize import minimize, Bounds, LinearConstraint

# 1. 目标函数
def objective(x):
    x1, x2 = x
    profit = 4*x1 - 0.3*x1**2 + 5*x2
    return -profit  # scipy minimize 函数默认是最小化问题，因此我们取负号来最大化利润

# 2. 初始解
x0 = np.array([0.0, 0.0]) # 非线性规划往往局部最优解不唯一，初始值的选择严重影响优化结果。

# 3. 变量边界
bounds = Bounds([0, 0], [None, None]) # 这里表示两者下限都是0，无上界

# 4. 线性约束
A = np.array([[2, 1],
              [1, 3]])
ub = np.array([10, 15])
linear_constraints = LinearConstraint(A, -np.inf, ub)

# 5. 求解
result = minimize(
    objective,
    x0,
    method='SLSQP',
    bounds=bounds,
    constraints=[linear_constraints]
)

# 6. 输出结果
print("求解成功:", result.success)
print("最优方案: x1 =", round(result.x[0], 4), ", x2 =", round(result.x[1], 4))
print("最大利润:", round(-result.fun, 4))
```
