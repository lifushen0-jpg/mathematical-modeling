## 我们将使用pulp库进行线性规划的学习

### 数学规划  
数学规划的目的是在一些限制条件下给出目标函数的最大值或最小值的方法  
在本节中，目标函数和限制条件将为线性函数

- 例1  

问题描述：
某工厂生产 A、B 两种产品，均需经过机器1和机器2加工。

每个 A 产品需机器1加工 2 小时，机器2加工 1 小时。

每个 B 产品需机器1加工 1 小时，机器2加工 3 小时。

机器1每天可用 10 小时，机器2每天可用 15 小时。

每件 A 产品利润 4 万元，每件 B 产品利润 5 万元。

求每天生产多少 A 和 B，使总利润最大？

```python
import pulp

# 1. 创建线性规划问题（求最大值）

model_lp = pulp.LpProblem("Linear_Programming", pulp.LpMaximize)  
# 此处的第一个变量为问题名称，无意义；第二个变量为求解目标，pulp.LpMaximize表示最大化问题，pulp.LpMinimize表示最小化问题    

# 2. 定义连续变量（下界默认为0）

x1 = pulp.LpVariable("x1", lowBound=0, cat='Continuous')
x2 = pulp.LpVariable("x2", lowBound=0, cat='Continuous')
# 此处如果想要同时设定一个变量的上下界，只需要x = pulp.LpVariable("x",lowBound=0,upBound=100)

# 3. 目标函数

model_lp += 4 * x1 + 5 * x2, "Total_Profit"

# 4. 约束条件

model_lp += 2 * x1 + x2 <= 10, "Machine1_Hours"
model_lp += x1 + 3 * x2 <= 15, "Machine2_Hours"

# 5. 求解

model_lp.solve()


# 6. 输出结果

print("========== 线性规划结果 ==========")
print(f"求解状态: {pulp.LpStatus[model_lp.status]}")
print(f"最优方案: x1 = {x1.value():.4f}, x2 = {x2.value():.4f}")
print(f"最大利润: {pulp.value(model_lp.objective):.4f}")
```

- 在实际问题中，我们可能遇到强制要求结果必须是整数的情况，例如某些产品如汽车，那么就需要进行整数线性规划。使用pulp可以很方便的进行整数线性规划,只需要在设置变量时使用整数变量即可

```python
import pulp

# 1. 创建线性规划问题（求最大值）

model_lp = pulp.LpProblem("Linear_Programming", pulp.LpMaximize)  
# 此处的第一个变量为问题名称，无意义；第二个变量为求解目标，pulp.LpMaximize表示最大化问题，pulp.LpMinimize表示最小化问题    

# 2. 定义整数变量（下界默认为0）

x1 = pulp.LpVariable("x1", lowBound=0, cat='integer')
x2 = pulp.LpVariable("x2", lowBound=0, cat='integer')
# 此处如果想要同时设定一个变量的上下界，只需要x = pulp.LpVariable("x",lowBound=0,upBound=100)

# 3. 目标函数

model_lp += 4 * x1 + 5 * x2, "Total_Profit"

# 4. 约束条件

model_lp += 2 * x1 + x2 <= 10, "Machine1_Hours"
model_lp += x1 + 3 * x2 <= 15, "Machine2_Hours"

# 5. 求解

model_lp.solve()


# 6. 输出结果

print("========== 线性规划结果 ==========")
print(f"求解状态: {pulp.LpStatus[model_lp.status]}")
print(f"最优方案: x1 = {x1.value()}, x2 = {x2.value()}")
print(f"最大利润: {pulp.value(model_lp.objective):.4f}")
```
