- eval() 的作用为将输入的字符串当作python的表达式计算并输出

```python
result = eval("3 + 4 * 5")
print(result)   
```

- map(function,iterable) 的作用为将函数(function)作用在可迭代对象(iterable)中的每个元素上

```python
list_str = ['1.0',' 2.0', '3.0']
list_float_1 = [float(i) for i in list_str]
list_float_2 = list(map(float, list_str))
print(list_float_1,list_float_2)
```

### 求解函数最小值

```python
from scipy.optimize import minimize_scalar

def f(x):
    return (x - 3) ** 2 + 2

result = minimize_scalar(f)
print(result.x)
print(result.fun)
```
